```python
df_airbnb['checkout'] = df_airbnb['checkout'].fillna("-")
```

```python
df_airbnb = df_airbnb.drop_duplicates()

df_airbnb['country'] = df_airbnb['country'].astype(str).str.strip()
```

```python
#Questions for the project?

#1 Which countries are most represented on the platform, and how does the average price vary among them?

df_places_country = df_airbnb.groupby('country')['id'].count().reset_index(name = 'number_places')

df_places_country = df_places_country.sort_values(by = 'number_places', ascending =False) #India, Italy and Greece are Top3

top_3_countries = df_places_country.head(3)

display(top_3_countries)

df_avg_price = df_airbnb.groupby('country')['price'].mean().astype(int).reset_index(name = 'avg_price')


df_avg_price = df_avg_price.sort_values(by = 'avg_price', ascending = True)


#Top5 cheapest countries by avg price
display(df_avg_price.head(5))


#Top5 most expensive countries by avg price
display(df_avg_price.tail(5))
```

```python
df_country_analysis = (
    df_airbnb
    .groupby('country')
    .agg(
        number_places=('id', 'count'),
        avg_price=('price', lambda x: x.mean().round().astype(int))
    )
    .reset_index()
    .sort_values(by='number_places', ascending=False)
)

display(df_country_analysis)
```

```python
#2 What is the difference in average price between properties that offer amenities like a pool or pet-friendly policies versus those that do not?

df_airbnb['amenities'] = df_airbnb['amenities'].astype(str)

filter_pool_petfriend = df_airbnb['amenities'].str.contains('Pool|Pet', case = False, na = False)

pool_petfriend_avg_price = df_airbnb[filter_pool_petfriend]['price'].mean().round().astype(int)
nopool_nopetfriend_avg_price = df_airbnb[~filter_pool_petfriend]['price'].mean().round().astype(int)

print(f"Average price for places with pool or petfriendly is {pool_petfriend_avg_price} in face of other places with average {nopool_nopetfriend_avg_price}")
```

```python
price_diff = pool_petfriend_avg_price - nopool_nopetfriend_avg_price
pct_diff = round((price_diff / nopool_nopetfriend_avg_price) * 100, 1)

print(f"Absolute difference: {price_diff} ({pct_diff}% higher)")
```

```python
#3 Does the rating score have significant impact on price?

#convert rating for a number

df_airbnb['rating_number'] = pd.to_numeric(df_airbnb['rating'], errors = 'coerce')

#correlation between rating and price

correlation = df_airbnb['rating_number'].corr(df_airbnb['price'])

#create rating groups to avaliate avg price and median

bins = [0,4.5, 4.8, 5.0]
labels = ['Below 4.5', '4.5-4.8', '4.8-5.0']
df_airbnb['rating_group'] = pd.cut(df_airbnb['rating_number'], bins=bins, labels=labels)

price_by_rating = (
    df_airbnb
    .groupby('rating_group', observed = False)['price']
    .agg(
        total_listings='count',
        avg_price=lambda x: x.mean().round().astype(int),
        median_price=lambda x: x.median()
    )
    .reset_index()
)

display(price_by_rating)

##Key Finding on Ratings vs. Price:
##At first glance, listings with the lowest ratings (< 4.0) appear to have a significantly higher average price (~25,026).
##However, deeper inspection reveals this is heavily affected by extreme outliers, creating some distortion in the analysis.
##Comparing median prices, evidence shows that low-rated properties perform quite similarly to top-rated ones, confirming that review scores do not dictate daily rates.
```

```python
#4 What is the proportion of listings managed by multi-property hosts versus single-property hosts, and do their pricing strategies differ?

host_count = df_airbnb['host_id'].value_counts() ##counts how many times a host appears

multi_host = host_count[host_count > 1].index

df_airbnb['host_type'] = np.where(df_airbnb['host_id'].isin(multi_host), 'Multi-property Host', 'Single-property Host')

host_comparison = (
    df_airbnb
    .groupby('host_type')
    .agg(
        total_listings=('id', 'count'),
        percentage=('id', lambda x: round((len(x) / len(df_airbnb)) * 100, 1)),
        avg_price=('price', lambda x: int(round(x.mean()))),
        median_price=('price', lambda x: int(round(x.median())))
    )
    .reset_index()
)

display(host_comparison)

#Although multi-property hosts hold higher average price, the median of the price and their share in the listings describe a different approach,
#much more concentrated on single-property hosts that present higher median price. In general, average prices do not differ so much which may sign a more open negotiation point (more competitive volume-driven pricing) with multiple hosts.
```

```python
#5 How does the 'cost per guest' metric behave as the capacity (bedrooms and allowed guests) increases?

# 1. Ensure valid guest count and engineer 'price_per_guest'
df_valid = df_airbnb[df_airbnb['guests'] > 0].copy()
df_valid['price_per_guest'] = df_valid['price'] / df_valid['guests']

# 2. Cost per guest grouped by allowed guests
guest_cost_summary = (
    df_valid
    .groupby('guests')
    .agg(
        total_listings=('id', 'count'),
        avg_price_per_guest=('price_per_guest', lambda x: int(round(x.mean()))),
        median_price_per_guest=('price_per_guest', lambda x: int(round(x.median()))),
        avg_total_price=('price', lambda x: int(round(x.mean())))
    )
    .reset_index()
)

# 3. Cost per guest grouped by bedrooms
bedroom_cost_summary = (
    df_valid
    .groupby('bedrooms')
    .agg(
        total_listings=('id', 'count'),
        avg_price_per_guest=('price_per_guest', lambda x: int(round(x.mean()))),
        median_price_per_guest=('price_per_guest', lambda x: int(round(x.median()))),
        avg_total_price=('price', lambda x: int(round(x.mean())))
    )
    .reset_index()
)

display(guest_cost_summary.head(8))
display(bedroom_cost_summary.head(8))


##Economies of Scale in Groups: listings for 3–6 guests delivers the highest cost efficiency per person, reducing median cost per guest from 2,727 (couples) to 1,624–1,974 (30–40% savings).
##Large Property: Expanding bedroom capacity past 4 bedrooms increases cost per guest up to 3,113, as listings transition from standard apartments to high-end villas.
```
